### 1. Area & Perimeter of a circle

```python
import math

def area_of_circle(r):
    return math.pi * r * r

def perimeter_of_circle(r):
    return 2 * math.pi * r

print(area_of_circle(5))
print(perimeter_of_circle(5))
```

### 2. add_all_nums

```python
def add_all_nums(*args):
    total = 0
    for item in args:
        if isinstance(item, (int, float)):
            total += item
        else:
            print(f"Warning: '{item}' is not a number and was skipped.")
    return total

print(add_all_nums(1, 2, 3, 'x', 4))
```

### 3. Celsius to Fahrenheit

```python
def convert_celsius_2_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

print(convert_celsius_2_fahrenheit(25))
```

### 4. check_season

```python
def check_season(month):
    month = month.lower()
    autumn = ['september', 'october', 'november']
    winter = ['december', 'january', 'february']
    spring = ['march', 'april', 'may']
    summer = ['june', 'july', 'august']

    if month in autumn:
        return 'Autumn'
    elif month in winter:
        return 'Winter'
    elif month in spring:
        return 'Spring'
    elif month in summer:
        return 'Summer'
    else:
        return 'Invalid month'

print(check_season('July'))
```

### 5. calculate_slope

```python
def calculate_slope(x1, y1, x2, y2):
    if x2 - x1 == 0:
        return "Undefined slope (vertical line)"
    return (y2 - y1) / (x2 - x1)

print(calculate_slope(1, 2, 3, 8))
```

### 6. solve_quadratic_eqn

```python
import math

def solve_quadratic_eqn(a, b, c):
    if a == 0:
        return "Not a quadratic equation (a cannot be 0)"

    discriminant = (b ** 2) - (4 * a * c)

    if discriminant > 0:
        x1 = (-b + math.sqrt(discriminant)) / (2 * a)
        x2 = (-b - math.sqrt(discriminant)) / (2 * a)
        return (x1, x2)
    elif discriminant == 0:
        x = -b / (2 * a)
        return (x,)
    else:
        real = -b / (2 * a)
        imag = math.sqrt(-discriminant) / (2 * a)
        return (f"{real}+{imag}j", f"{real}-{imag}j")

print(solve_quadratic_eqn(1, -3, 2))
```

### 7. print_list

```python
def print_list(lst):
    for item in lst:
        print(item)

print_list([1, 2, 3])
```

### 9. capitalize_list_items

```python
def capitalize_list_items(lst):
    return [str(item).capitalize() for item in lst]

print(capitalize_list_items(['potato', 'tomato', 'mango']))
```

### 10. add_item

```python
def add_item(lst, item):
    lst.append(item)
    return lst

food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(add_item(food_staff, 'Fungi'))

numbers = [2, 3, 7, 9]
print(add_item(numbers, 5))
```

### 11. remove_item

```python
def remove_item(lst, item):
    lst.remove(item)
    return lst

food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(remove_item(food_staff, 'Mango'))

numbers = [2, 3, 7, 9]
print(remove_item(numbers, 3))
```

### 12. sum_of_numbers

```python
def sum_of_numbers(n):
    return sum(range(1, n + 1))

print(sum_of_numbers(5))
print(sum_of_numbers(10))
print(sum_of_numbers(100))
```

### 13. sum_of_odds

```python
def sum_of_odds(n):
    return sum(num for num in range(1, n + 1) if num % 2 != 0)

print(sum_of_odds(10))
```

### 14. sum_of_even

```python
def sum_of_even(n):
    return sum(num for num in range(1, n + 1) if num % 2 == 0)

print(sum_of_even(10))
```
